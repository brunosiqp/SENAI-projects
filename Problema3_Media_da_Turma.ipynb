## Problema 3 - Média da Turma

**Aluno: Bruno Siqueira**

O programa recebe as notas de cinco alunos utilizando uma estrutura de repetição, calcula a média da turma e informa se o aluno 1 foi aprovado ou reprovado. O aluno 1 é aprovado quando sua nota é igual ou superior à média da turma.


### Resolução

Foi utilizada a estrutura `for` para solicitar as cinco notas. A estrutura `while` valida cada entrada e impede notas fora do intervalo de 0 a 100.


```python
notas = []

for aluno in range(1, 6):
    while True:
        try:
            nota = float(input(f"Digite a nota do aluno {aluno}: ").replace(",", "."))
            if 0 <= nota <= 100:
                notas.append(nota)
                break
            print("A nota deve estar entre 0 e 100.")
        except ValueError:
            print("Digite uma nota válida.")

media_turma = sum(notas) / len(notas)

print(f"\nMédia da Turma: {media_turma:g}")

if notas[0] >= media_turma:
    print("Portanto, o aluno 1 está APROVADO!")
else:
    print("Portanto, o aluno 1 está REPROVADO!")
```

### Exemplo esperado

Para as notas `60`, `40`, `30`, `80` e `90`, a média é `60`. Como a primeira nota é igual à média, o aluno 1 está aprovado.

O algoritmo lê o arquivo `mailbox.txt`, extrai o endereço de e-mail de cada linha iniciada por `From `, elimina repetições e grava a listagem no arquivo `output.txt`.


### Arquivo necessário

Mantenha `mailbox.txt` na mesma pasta do notebook. No Google Colab, envie o arquivo antes de executar o código.


```python
try:
    from google.colab import files
    enviados = files.upload()
    print("Arquivos enviados:", list(enviados.keys()))
except ImportError:
    print("Execução local: usando os arquivos da pasta atual.")
```

```python
from pathlib import Path

pasta = Path(__file__).resolve().parent if '__file__' in globals() else Path.cwd()
arquivo_entrada = pasta / 'mailbox.txt'
arquivo_saida = pasta / 'output.txt'

emails = []

with arquivo_entrada.open('r', encoding='utf-8', errors='replace') as arquivo:
    for linha in arquivo:
        if linha.startswith('From '):
            partes = linha.split()
            if len(partes) > 1:
                email = partes[1]
                if email not in emails:
                    emails.append(email)

with arquivo_saida.open('w', encoding='utf-8') as arquivo:
    for email in emails:
        arquivo.write(email + '\n')

print(f'Quantidade de e-mails únicos: {len(emails)}')
print(f'Arquivo gerado: {arquivo_saida.name}')
print('\nE-mails extraídos:')
for email in emails:
    print(email)
```

### Resultado

Após a execução, será criado o arquivo `output.txt`, contendo um endereço por linha e sem repetições.

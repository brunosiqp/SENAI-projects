O algoritmo simula dez lançamentos de um dado, armazena os resultados em uma lista e calcula a média, a quantidade de valores acima da média, o maior valor e o menor valor.


### Lógica utilizada

- `random.randint(1, 6)` gera um número inteiro aleatório entre 1 e 6.
- Cada lançamento é armazenado na lista `lancamentos`.
- A média é calculada por `sum(lancamentos) / len(lancamentos)`.
- A quantidade acima da média é calculada percorrendo os valores armazenados.
- `max()` e `min()` determinam, respectivamente, o maior e o menor valor da lista.


```python
import random

lancamentos = []

for _ in range(10):
    valor = random.randint(1, 6)
    lancamentos.append(valor)

media = sum(lancamentos) / len(lancamentos)
quantidade_acima_media = sum(1 for valor in lancamentos if valor > media)
maior_valor = max(lancamentos)
menor_valor = min(lancamentos)

print(f"Valores dos lançamentos: {lancamentos}")
print(f"Média dos valores: {media:.1f}")
print(f"Quantidade de valores acima da média: {quantidade_acima_media}")
print(f"Maior valor: {maior_valor}")
print(f"Menor valor: {menor_valor}")
```

### Observação

Como os lançamentos são aleatórios, os valores e os resultados calculados podem mudar a cada execução do notebook.

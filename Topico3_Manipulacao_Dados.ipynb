# 3.1 Entrada e Saída (E/S) de dados

## Leitura de arquivos do tipo texto (.txt)

A função `open()` abre o arquivo no modo indicado. O argumento `r` representa leitura (*read*). Após a utilização, o arquivo deve ser fechado com `close()`.

```python
arquivo1 = open(pasta / 'file1.txt', 'r', encoding='utf-8-sig')
print(arquivo1)
arquivo1.close()
```

O objeto exibido é um identificador do arquivo. Para acessar o conteúdo, percorremos suas linhas.

```python
arquivo1 = open(pasta / 'file1.txt', 'r', encoding='utf-8-sig')
for linha in arquivo1:
    print(linha, end='')
arquivo1.close()
```

Outra forma é utilizar a estrutura `with`. Ao final do bloco, o arquivo é fechado automaticamente.

```python
with open(pasta / 'file1.txt', 'r', encoding='utf-8-sig') as arquivo1:
    for linha in arquivo1.readlines():
        print(linha, end='')
```

## Gravação de arquivo texto

O modo `w` representa gravação (*write*). Caso o arquivo já exista, seu conteúdo será sobrescrito.

```python
with open(pasta / 'fileXYZ.txt', 'w', encoding='utf-8') as arquivo2:
    for linha in range(1, 21):
        arquivo2.write(f'Linha: {linha}\n')

print('Arquivo gravado com sucesso:', pasta / 'fileXYZ.txt')
```

```python
with open(pasta / 'fileXYZ.txt', 'r', encoding='utf-8') as arquivo2:
    print(arquivo2.read())
```

## Arquivos delimitados

O arquivo `movies.csv` utiliza ponto e vírgula como delimitador. Primeiro, o conteúdo pode ser lido como texto comum.

```python
arquivo_csv = open(pasta / 'movies.csv', 'r', encoding='utf-8-sig')
for indice, linha in enumerate(arquivo_csv):
    print(linha, end='')
    if indice == 5:  # limita a exibição ao cabeçalho e cinco registros
        break
arquivo_csv.close()
```

Para facilitar a manipulação dos dados, utilizamos `read_csv()` da biblioteca `pandas`, informando `sep=';'`.

```python
import pandas as pd

df = pd.read_csv(pasta / 'movies.csv', sep=';')
df.head()
```

```python
print('Quantidade de linhas:', df.shape[0])
print('Quantidade de colunas:', df.shape[1])
print('Colunas do conjunto de dados:')
print(df.columns.tolist())
```

O arquivo `movies2.txt` contém os mesmos tipos de colunas, mas utiliza tabulação como delimitador. Nesse caso, usamos `sep='\t'`.

```python
df_txt = pd.read_csv(pasta / 'movies2.txt', sep='\t')
df_txt.head()
```

```python
print('Quantidade de linhas no CSV:', len(df))
print('Quantidade de linhas no TXT:', len(df_txt))
print('As colunas são iguais?', list(df.columns) == list(df_txt.columns))
```

## Arquivos JSON

O módulo `json` permite converter o conteúdo do arquivo JSON em objetos Python.

```python
import json

with open(pasta / 'example.json', 'r', encoding='utf-8-sig') as arquivo_json:
    dados_json = json.load(arquivo_json)

print(dados_json)
print('Tipo:', type(dados_json))
```

```python
for item in dados_json:
    print(f"a = {item['a']}, b = {item['b']}, c = {item['c']}")
```

# 3.2 Manipulação de Strings

O exemplo abaixo segue o padrão apresentado no material, utilizando uma cidade e uma UF separadas por vírgula.

```python
cidade_estado = 'Blumenau,SC'

maiuscula = cidade_estado.upper()
print('Maiúscula:', maiuscula)

minuscula = cidade_estado.lower()
print('Minúscula:', minuscula)

caracteres = len(cidade_estado)
print('Quantidade de caracteres:', caracteres)

espacos = cidade_estado.count(' ')
print('Quantidade de espaços:', espacos)

cidade_estado = cidade_estado.strip()
print('Sem espaços nas extremidades:', cidade_estado)
```

```python
print('Primeiros 5 caracteres:', cidade_estado[0:5])

posicao = cidade_estado.find(',')
print('Posição da vírgula:', posicao)

cidade = cidade_estado[0:posicao]
print('Cidade:', cidade.rstrip())

uf = cidade_estado[posicao + 1:]
print('UF:', uf)

uf2 = cidade_estado[-2:]
print('UF pelos dois últimos caracteres:', uf2)

print('Forma reversa:', cidade_estado[::-1])
```

# 3.3 Tipos de dados temporais (Data/Hora)

O módulo `datetime` disponibiliza classes para representar datas, horários e intervalos de tempo.

```python
from datetime import datetime, date, timedelta

agora = datetime.now()
hoje = date.today()

print('Data e hora atuais:', agora)
print('Data atual:', hoje)
print('Data formatada:', agora.strftime('%d/%m/%Y %H:%M:%S'))
```

```python
data_texto = '15/10/2026'
data_convertida = datetime.strptime(data_texto, '%d/%m/%Y')
print('Data convertida:', data_convertida)
print('Tipo:', type(data_convertida))
```

```python
amanha = hoje + timedelta(days=1)
proxima_semana = hoje + timedelta(days=7)

print('Amanhã:', amanha.strftime('%d/%m/%Y'))
print('Daqui a 7 dias:', proxima_semana.strftime('%d/%m/%Y'))
print('Diferença em dias:', (proxima_semana - hoje).days)
```

## Exercício 1
Leia o arquivo `file1.txt`, armazene suas linhas em uma lista e apresente a quantidade de linhas e palavras.

```python
with open(pasta / 'file1.txt', 'r', encoding='utf-8-sig') as arquivo:
    linhas = arquivo.readlines()

quantidade_linhas = len(linhas)
quantidade_palavras = sum(len(linha.split()) for linha in linhas)

print('Quantidade de linhas:', quantidade_linhas)
print('Quantidade de palavras:', quantidade_palavras)
```

Apresente a média da nota IMDb e a média de duração dos filmes.

### Análise com Pandas e NumPy

O arquivo de dados não possui cabeçalho. Por isso, os 14 campos são nomeados conforme a documentação: identificador do tipo e 13 constituintes químicos, começando pelo teor alcoólico.


```python
from pathlib import Path
import numpy as np
import pandas as pd

COLUNAS = [
    "Tipo", "Teor alcoólico", "Ácido málico", "Cinzas",
    "Alcalinidade das cinzas", "Magnésio", "Fenóis totais",
    "Flavanoides", "Fenóis não flavonoides", "Proantocianinas",
    "Intensidade da cor", "Tonalidade", "OD280/OD315", "Prolina"
]

pasta = Path(__file__).resolve().parent if "__file__" in globals() else Path.cwd()
arquivo_dados = pasta / "winedata.txt"
arquivo_nomes = pasta / "winenames.txt"

if not arquivo_dados.exists():
    raise FileNotFoundError("O arquivo winedata.txt deve estar na mesma pasta do programa.")
if not arquivo_nomes.exists():
    raise FileNotFoundError("O arquivo winenames.txt deve estar na mesma pasta do programa.")

# Leitura dos dados originais fornecidos na atividade.
df = pd.read_csv(arquivo_dados, header=None, names=COLUNAS)

# Leitura do arquivo descritivo para comprovar que ele também está disponível.
descricao = arquivo_nomes.read_text(encoding="utf-8", errors="replace")

# a) Quantidade de vinhos por tipo.
quantidade_por_tipo = df["Tipo"].value_counts().sort_index()

# b) Maior teor alcoólico dos vinhos.
maior_teor = np.max(df["Teor alcoólico"].to_numpy())

# c) Menor teor alcoólico dos vinhos.
menor_teor = np.min(df["Teor alcoólico"].to_numpy())

# d) Média de teor alcoólico por tipo de vinho.
media_por_tipo = df.groupby("Tipo")["Teor alcoólico"].mean()

print("=== Problema 11 - Análise do conjunto Wine ===")
print(f"Arquivo de dados: {arquivo_dados.name}")
print(f"Arquivo descritivo: {arquivo_nomes.name}")
print(f"Registros carregados: {len(df)}")
print(f"Descrição carregada: {len(descricao)} caracteres")

print("\na) Quantidade de vinhos por tipo:")
for tipo, quantidade in quantidade_por_tipo.items():
    print(f"Tipo {tipo}: {quantidade} vinhos")

print(f"\nb) Maior teor alcoólico: {maior_teor:.2f}")
print(f"c) Menor teor alcoólico: {menor_teor:.2f}")

print("\nd) Média de teor alcoólico por tipo de vinho:")
for tipo, media in media_por_tipo.items():
    print(f"Tipo {tipo}: {media:.2f}")
```

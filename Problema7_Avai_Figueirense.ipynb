O programa recebe os resultados de cinco jogos, armazena cada placar em um dicionário e calcula todas as informações solicitadas com base nesses dados.


### Resolução

Foram utilizadas funções para contar vitórias, contar empates e calcular o saldo de gols. Cada partida é armazenada no dicionário `resultados_jogos`, com os gols de Avaí e Figueirense.


```python

def contar_vitorias(jogos, time1, time2):
    vitorias = 0
    for placar in jogos.values():
        if placar[time1] > placar[time2]:
            vitorias += 1
    return vitorias


def contar_empates(jogos):
    empates = 0
    for placar in jogos.values():
        if placar["Avaí"] == placar["Figueirense"]:
            empates += 1
    return empates


def calcular_saldo_gols(jogos, time1, time2):
    saldo = 0
    for placar in jogos.values():
        saldo += placar[time1] - placar[time2]
    return saldo


resultados_jogos = {}

for numero_jogo in range(1, 6):
    while True:
        try:
            resultado = input(
                f"Digite o resultado do jogo {numero_jogo} "
                "(Avaí x Figueirense), no formato 2x1: "
            ).lower().replace(" ", "")

            gols_avai, gols_figueirense = resultado.split("x")
            gols_avai = int(gols_avai)
            gols_figueirense = int(gols_figueirense)

            if gols_avai < 0 or gols_figueirense < 0:
                raise ValueError

            resultados_jogos[f"Jogo {numero_jogo}"] = {
                "Avaí": gols_avai,
                "Figueirense": gols_figueirense,
            }
            break
        except ValueError:
            print("Resultado inválido. Digite no formato 2x1, usando valores inteiros não negativos.")

print("\n=== Resultados dos Jogos ===")
for jogo, placar in resultados_jogos.items():
    print(
        f"{jogo} (Avaí x Figueirense): "
        f"{placar['Avaí']} x {placar['Figueirense']}"
    )

vitorias_avai = contar_vitorias(resultados_jogos, "Avaí", "Figueirense")
vitorias_figueirense = contar_vitorias(resultados_jogos, "Figueirense", "Avaí")
empates = contar_empates(resultados_jogos)
saldo_avai = calcular_saldo_gols(resultados_jogos, "Avaí", "Figueirense")
saldo_figueirense = calcular_saldo_gols(resultados_jogos, "Figueirense", "Avaí")

print("\n=== Estatísticas ===")
print(f"Quantidade de vitórias do Avaí: {vitorias_avai}")
print(f"Quantidade de vitórias do Figueirense: {vitorias_figueirense}")
print(f"Quantidade de empates: {empates}")
print(f"Saldo de gols do Avaí: {saldo_avai}")
print(f"Saldo de gols do Figueirense: {saldo_figueirense}")
```

### Formato de entrada

Digite cada placar no formato `2x1`. Ao final das cinco entradas, o notebook exibirá a listagem completa, as vitórias de cada equipe, os empates e os saldos de gols.

A aplicação lê o código, a descrição e o preço inicial de um item. Em seguida, apresenta o preço do primeiro ao sétimo dia, aplicando uma redução diária de 10% sobre o preço do dia anterior.


### Lógica utilizada

O preço do primeiro dia corresponde ao valor informado pelo usuário. Após exibir o preço de cada dia, o valor é multiplicado por `0.90`, preservando 90% do preço anterior. A repetição é realizada com `for` para os sete dias da promoção.


```python
print("Olá Usuário, digite os dados do item.")
codigo = input("Digite o código do item: ").strip()
descricao = input("Digite a descrição do item: ").strip()

while True:
    try:
        preco_inicial = float(input("Digite o preço inicial do item: ").replace(",", "."))
        if preco_inicial > 0:
            break
        print("O preço deve ser maior que zero.")
    except ValueError:
        print("Digite um preço válido.")

print("\nInformações do Item:")
print(f"Código: {codigo}")
print(f"Descrição: {descricao}\n")

preco = preco_inicial
for dia in range(1, 8):
    print(f"Preço no dia {dia}: R$ {preco:.2f}")
    preco *= 0.90
```

### Exemplo do enunciado

Para o código `001`, descrição `Sapato Social` e preço inicial `399`, os valores apresentados do primeiro ao sétimo dia são: R$ 399,00; R$ 359,10; R$ 323,19; R$ 290,87; R$ 261,78; R$ 235,61; e R$ 212,04.

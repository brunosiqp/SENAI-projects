O programa gera 15 números inteiros aleatórios entre 1 e 100 e calcula o quadrado da diferença de cada número em relação à média dos 15 valores.


### Resolução

A lista de números e a lista dos quadrados das diferenças são criadas com *list comprehensions*. A média é calculada pela soma dos valores dividida pela quantidade de elementos.


```python
import random

# Gera 15 números inteiros aleatórios entre 1 e 100 usando list comprehension.
numeros = [random.randint(1, 100) for _ in range(15)]

# Calcula a média dos 15 valores.
media = sum(numeros) / len(numeros)

# Calcula o quadrado da diferença de cada valor em relação à média.
quadrados_diferenca = [round((numero - media) ** 2, 2) for numero in numeros]

print("Números aleatórios:", numeros)
print(f"Média: {media:.2f}")
print("Quadrados da diferença em relação à média:", quadrados_diferenca)
```

### Observação

Os valores mudam a cada execução porque são gerados aleatoriamente. Os quadrados das diferenças são arredondados para duas casas decimais.

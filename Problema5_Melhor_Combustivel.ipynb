O algoritmo recebe os preços por litro da gasolina e do etanol e utiliza uma função para indicar a opção mais vantajosa.


### Regras utilizadas

- Os dois preços devem estar entre R$ 0,01 e R$ 10,00. Caso contrário, o resultado será `PREÇO INVÁLIDO`.
- A relação é calculada dividindo o preço do etanol pelo preço da gasolina.
- Se a relação for menor ou igual a 73%, a melhor opção será `ETANOL`.
- Se a relação for superior a 73%, a melhor opção será `GASOLINA`.


```python

def melhor_combustivel(preco_gasolina, preco_etanol):
    """Retorna GASOLINA, ETANOL ou PREÇO INVÁLIDO."""
    if not (0.01 <= preco_gasolina <= 10.00) or not (0.01 <= preco_etanol <= 10.00):
        return "PREÇO INVÁLIDO"

    relacao = preco_etanol / preco_gasolina

    if relacao <= 0.73:
        return "ETANOL"
    return "GASOLINA"


print("Bem-vindo à Calculadora do melhor combustível!")

try:
    gasolina = float(input("Digite o preço da Gasolina em R$: ").replace(",", "."))
    etanol = float(input("Digite o preço do Etanol em R$: ").replace(",", "."))

    resultado = melhor_combustivel(gasolina, etanol)

    if resultado == "PREÇO INVÁLIDO":
        print("PREÇO INVÁLIDO")
    else:
        print(f"Para os preços informados, a melhor opção é a utilização da {resultado}.")
except ValueError:
    print("PREÇO INVÁLIDO")
```

### Exemplo do enunciado

Para gasolina a R$ 3,50 e etanol a R$ 2,75, a relação é aproximadamente 78,57%. Como o resultado é superior a 73%, a melhor opção é a gasolina.

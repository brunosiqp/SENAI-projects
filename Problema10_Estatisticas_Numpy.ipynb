O programa recebe os resultados de cinco partidas entre Avaí e Figueirense e utiliza apenas um array NumPy para armazenar os placares e calcular as estatísticas.


### Estrutura adotada

O array possui cinco linhas e duas colunas. A coluna 0 armazena os gols do Avaí e a coluna 1 armazena os gols do Figueirense. As vitórias e os empates são calculados com comparações vetorizadas, enquanto os saldos são calculados com `np.sum()`.


```python
import numpy as np

# Array NumPy 5x2: coluna 0 = Avaí; coluna 1 = Figueirense.
resultados = np.zeros((5, 2), dtype=int)

for jogo in range(5):
    while True:
        try:
            placar = input(
                f"Digite o resultado do jogo {jogo + 1} "
                "(Avaí x Figueirense): "
            ).upper().replace(" ", "")

            gols_avai, gols_figueirense = placar.split("X")
            gols_avai = int(gols_avai)
            gols_figueirense = int(gols_figueirense)

            if gols_avai < 0 or gols_figueirense < 0:
                raise ValueError

            resultados[jogo] = [gols_avai, gols_figueirense]
            break
        except ValueError:
            print("Resultado inválido. Digite no formato 2x1.")

print("\n=== Resultados dos Jogos ===")
for jogo in range(resultados.shape[0]):
    print(
        f"Jogo {jogo + 1} (Avaí x Figueirense): "
        f"{resultados[jogo, 0]} x {resultados[jogo, 1]}"
    )

# Todos os cálculos estatísticos são realizados com operações NumPy.
vitorias_avai = np.count_nonzero(resultados[:, 0] > resultados[:, 1])
vitorias_figueirense = np.count_nonzero(resultados[:, 1] > resultados[:, 0])
empates = np.count_nonzero(resultados[:, 0] == resultados[:, 1])
saldo_avai = np.sum(resultados[:, 0] - resultados[:, 1])
saldo_figueirense = np.sum(resultados[:, 1] - resultados[:, 0])

print("\n=== Estatísticas ===")
print(f"Vitórias do Avaí: {vitorias_avai}")
print(f"Vitórias do Figueirense: {vitorias_figueirense}")
print(f"Empates: {empates}")
print(f"Saldo de gols do Avaí: {saldo_avai}")
print(f"Saldo de gols do Figueirense: {saldo_figueirense}")
```

### Exemplo do enunciado

Para os placares `1x1`, `2x2`, `7x1`, `2x5` e `1x2`, o resultado é: uma vitória do Avaí, duas do Figueirense, dois empates, saldo 2 para o Avaí e -2 para o Figueirense.
